```python
# Voeg hier je plots in.

Frequenties= np.array([5.4,7.2,7.5,4.1,8.4,10.2,6.5])
Frequentessorted= np.sort(Frequenties)*2*np.pi
Amplitudes= np.array([0.19/1.12,0.12/1.81,0.23/2.8,0.43/4.85,0.34/5.82,0.41/6.08,0.26/6.31])

def func(w,omega):
    return np.abs(1/(1-((w)**2/omega**2)))

wplot=np.linspace(0,20,1000)


def func3(g,dx):
    return np.sqrt(g/dx)

omega_r = func3(g,0.20)

plt.loglog(wplot,func(wplot,Kantelfreuquentie),label='theoretische data')
plt.loglog(wplot,func(wplot,omega_r), label= 'gemeten data')
plt.loglog(Frequentessorted,Amplitudes, 'o',label= 'gemeten data')
plt.xlabel('log Frequentie (Hz)')
plt.ylabel('log|Aout/Ain| ')
plt.legend()
plt.show()
```
